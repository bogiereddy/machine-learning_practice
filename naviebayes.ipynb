```python
import pandas as pd
df =pd.read_csv('/content/emails.csv')
```

```python
df
```

```python
df.groupby('spam').describe()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test, y_train, y_test = train_test_split(df.text,df.spam, test_size=0.20, random_state=42)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
v = CountVectorizer()#converting text to lowercase, ignoring punctuation
X_train_count = v.fit_transform(X_train.values) # builds a vocabulary of all unique words, and converts the text into a matrix of word counts.
X_train_count.toarray()[:3]#the first 3 documents/rows.
```

```python
from sklearn.naive_bayes import MultinomialNB
model = MultinomialNB()
model.fit(X_train_count,y_train)
```

```python
emails=['research and development charges.',
        'upto 20%discount on pricing, eclusive offer just for you. dont miss this reward!'
        ]
emails_count= v.transform(emails)
model.predict(emails_count)
```

```python
X_test_count = v.transform(X_test)
model.score(X_test_count,y_test)
```

```python
from sklearn.pipeline import Pipeline
clf = Pipeline([
                ('vectorizer',CountVectorizer()),
                ('nb',MultinomialNB())
])
```

```python
clf.fit(X_train,y_train)
```

```python
clf.score(X_test,y_test)
```

```python
clf.predict(emails)
```
