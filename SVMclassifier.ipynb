```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.svm import SVC
from sklearn.inspection import DecisionBoundaryDisplay

# 1. Load Iris data (Using first 2 features for 2D visualization)
iris = datasets.load_iris()
X = iris.data[:, :2]  # Sepal length, Sepal width
y = iris.target

# 2. Train the SVM Model (Using RBF kernel as an example)
svm_model = SVC(kernel='rbf', C=1.0, gamma='scale')
svm_model.fit(X, y)

# 3. Setup the plot
fig, ax = plt.subplots(figsize=(8, 6))

# 4. Plot the decision boundaries and decision regions
DecisionBoundaryDisplay.from_estimator(
    svm_model,
    X,
    response_method="predict",
    cmap=plt.cm.coolwarm,
    alpha=0.8,
    ax=ax,
    xlabel=iris.feature_names[0],
    ylabel=iris.feature_names[1],
)

# 5. Scatter plot of the real data points
scatter = ax.scatter(X[:, 0], X[:, 1], c=y, cmap=plt.cm.coolwarm, edgecolors="k")

# 6. Add labels and title
ax.set_title("SVM Decision Boundary on Iris Dataset (Sepal Features)")
ax.legend(scatter.legend_elements()[0], iris.target_names, loc="upper right", title="Classes")

plt.show()
```
